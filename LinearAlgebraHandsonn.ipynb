```python
import numpy as np
import pandas as pd
A = np.array ([[4,5,1],[3,8,4],[20,3,6]])
B = np.array ([[12,14,5],[45,5,5],[9,8,3]])
```

```python
A
```

```python
B
```

```python
# L.H.S
lhs =(A+B).T
print (lhs)
```

```python
#R.H.S
rhs = A.T + B.T
print (rhs)
```

```python
if np.array_equal (lhs,rhs):
    print('correct')
else:
    print('wrong')
```

```python
# MULTIPLICATIVE PROPERTY
```

```python
lhs=np.dot(A,B).T
print(lhs)
```

```python
rhs= B.T.dot(A.T)
rhs
```

```python
if np.array_equal (lhs,rhs):
    print('correct')
else:
    print('wrong')
```

```python
# check matrix is singular or invertible
```

```python
A= np.random.randint(1,10, size=(3,3))
A
```

```python
detA = np.linalg.det(A)
```

```python
if detA == 0:
    print('A is singular')
else:
    print('A is invertible')
```

```python
# INVERSE MATRIXX
```

```python
M = np.array ([[1,2],[3,4]])
M
```

```python
M_inv =np.linalg.inv(M)
M_inv
```

```python
result = np.dot(M,M_inv)
result
```

```python
if np.allclose(result, np.identity(2)):
    print("True")
else:
    print("False")
```
